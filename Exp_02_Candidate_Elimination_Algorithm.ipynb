```python
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

n = len(data[0]) - 1

S = ['0'] * n
G = [['?'] * n]

for example in data:
    x = example[:-1]
    label = example[-1]

    if label == 'Yes':
        for i in range(n):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

        G = [g for g in G if all(
            g[i] == '?' or g[i] == S[i] for i in range(n)
        )]

    else:
        new_G = []
        for g in G:
            for i in range(n):
                if g[i] == '?':
                    if S[i] != '?':
                        new_g = g.copy()
                        new_g[i] = S[i]
                        new_G.append(new_g)
        G = new_G

print("Specific Boundary:")
print(S)

print("\nGeneral Boundary:")
for g in G:
    print(g)
```
